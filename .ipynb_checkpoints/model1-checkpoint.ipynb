```python
# BLOCK 1: FAST TRAINING ON /Users/Apple/Desktop/project/dataset/images/train

!pip install -q ultralytics gradio opencv-python

import os
import cv2
from ultralytics import YOLO

# Paths
PROJECT_ROOT = "/Users/Apple/Desktop/project"
TRAIN_DIR = "/Users/Apple/Desktop/project/dataset/images/train"  # contains crack/ and no_crack/
RUN_NAME = "crack_classifier_fast"


def remove_corrupt_images(folder):
    for root, _, files in os.walk(folder):
        for f in files:
            if not f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp")):
                continue
            path = os.path.join(root, f)
            img = cv2.imread(path)
            if img is None:
                print("Removing corrupt image:", path)
                try:
                    os.remove(path)
                except Exception as e:
                    print("Could not remove", path, "->", e)


print("Cleaning corrupt images in:", TRAIN_DIR)
remove_corrupt_images(TRAIN_DIR)
print("Cleaning done.")

# Train YOLOv8 classification model (FAST: few epochs)
model = YOLO("yolov8s-cls.pt")

model.train(
    data=TRAIN_DIR,   # auto-splits into train_split/train and train_split/val
    epochs=50,         # make this bigger (e.g. 20, 50) for better accuracy
    batch=16,
    imgsz=224,
    workers=4,
    name=RUN_NAME,
    project=os.path.join(PROJECT_ROOT, "runs")  # ensures runs path = /Users/Apple/Desktop/project/runs
)

print("Training finished.")
```
